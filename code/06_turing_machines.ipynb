## Basic theory

A Turing machine is a mathematical model of computation. It consists of:

- a finite set of states,
- an input tape divided into cells,
- a read/write head that sits on a cell,
- a transition function that tells the machine what to do next.

At each step, the machine:

1. reads the symbol under the head,
2. chooses a transition based on the current state and symbol,
3. writes a new symbol,
4. moves left or right,
5. enters a new state.

The machine may continue indefinitely, or it may eventually halt.

## Why Turing machines matter

A finite automaton has no memory beyond the current state, and a PDA adds a stack. A Turing machine adds a tape, which is effectively an unbounded memory device. This gives it the power to perform arbitrary algorithmic computation, at least in the theoretical sense.

This is why the Turing machine is the standard model used to formalize computability and complexity.

## Important terminology and notation

- Tape: an infinite or effectively unbounded sequence of cells
- Head: the pointer that reads and writes on the tape
- Symbol: a character stored on the tape, often 0, 1, or blank
- State: the machine's internal control state
- Transition: a rule of the form (state, symbol) -> (new_state, new_symbol, direction)
- Halt: a stopping condition where the machine stops computing
- Accept/reject: a TM may accept or reject an input depending on the computation
- Blank symbol: a special tape symbol representing an empty cell

## A simple example: incrementing a binary number

Suppose the tape initially contains `101` and we want a machine that adds 1 to that binary number. The machine would move to the right end of the input, change the last 1 to 0 if needed, and carry over if the end is 1.

This is a direct example of a Turing machine performing a basic arithmetic task.

## Example of a transition rule

A transition may look like:

(state q0, symbol 1) -> (state q1, symbol 0, move right)

This means:

- if the machine is in state q0 and reads 1,
- it writes 0,
- moves right,
- and enters state q1.

This is the basic algebra of Turing-machine computation.

## Python representation of a TM

We will use a simple simulation in Python. The tape is a list of symbols, and the head position is an integer. The machine state and transition table determine the behavior.


```python
# A simple Turing machine that replaces the first 1 with 0 and halts.
# This is a toy example to show how tape, head, and states work.


def simulate_simple_tm(input_string):
    tape = list(input_string)
    head = 0
    state = 'q0'

    # One transition rule:
    # q0 reads '1' -> write '0', move right, go to qhalt
    while state != 'qhalt':
        if head >= len(tape):
            tape.append(' ')

        current = tape[head]

        if state == 'q0':
            if current == '1':
                tape[head] = '0'
                head += 1
                state = 'qhalt'
            elif current == '0':
                head += 1
            else:
                # Blank or other symbol: stop
                state = 'qhalt'

    return ''.join(tape).strip()


for word in ['101', '111', '000', '1', '011']:
    print(word, '->', simulate_simple_tm(word))

print("\nThis shows the basic idea: read, write, move, and update state.")
```

## Example 1: replacing a symbol

The machine above reads a symbol and writes a new symbol. This is the simplest form of TM computation: a machine can modify its tape.

If the input is `111`, then the first 1 is replaced by 0 and the machine halts. The tape becomes something like `011`.

## Example 2: moving across the tape

A more interesting machine moves right across the tape while changing symbols. This is how a TM can perform operations such as copying, marking, or checking whether the input has a certain pattern.

## Example 3: deciding a pattern

A Turing machine can decide whether a string is of a certain form by scanning the tape and using the state to remember what has been seen so far. For example, a machine can check whether there are as many 0s as 1s by going back and forth while marking matched pairs.

This is much more powerful than a finite automaton because the tape allows the machine to revisit earlier data.

## Python implementation: a tiny copy machine

The following machine copies a 1-bit pattern from one side of the tape to another. This is a simple example of using memory on the tape.


```python
# A tiny demonstration of TM-style rewriting on a tape.
# This example replaces every '1' with 'X' and marks the tape.


def mark_ones(word):
    tape = list(word)
    i = 0

    while i < len(tape):
        if tape[i] == '1':
            tape[i] = 'X'
        i += 1

    return ''.join(tape)


for word in ['111', '0101', '000', '10101']:
    print(word, '->', mark_ones(word))

print("\nA Turing machine can modify its tape by reading symbols and writing new ones.")
```
