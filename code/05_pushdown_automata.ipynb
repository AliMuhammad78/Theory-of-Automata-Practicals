We will study the basic definition, how the stack works, and how to simulate a PDA in Python for simple examples.

## Basic theory

A pushdown automaton is a finite automaton with a stack. A stack is a last-in, first-out memory structure.

This extra memory allows the machine to remember unmatched symbols, which is exactly what is needed for languages like:

- balanced parentheses: (())
- strings with equal numbers of a's and b's in a specific order
- palindromes in some simplified forms

A PDA can read input symbols one by one and, depending on the current state and top of stack, either:

- push a symbol onto the stack,
- pop a symbol from the stack,
- leave the stack unchanged.

This makes the PDA more powerful than a DFA, because a stack gives it unbounded memory in a structured way.

## Formal definition

A PDA is usually described as a tuple:

M = (Q, Σ, Γ, δ, q0, Z0, F)

where:

- Q is the set of states,
- Σ is the input alphabet,
- Γ is the stack alphabet,
- δ is the transition function,
- q0 is the start state,
- Z0 is the initial stack symbol,
- F is the set of accepting states.

In a simple simulation, we often use a stack represented as a Python list, where the last element of the list is the top of the stack.

## Acceptance modes

There are two common ways a PDA may accept input:

1. By final state: the machine accepts if it ends in an accepting state after reading the whole input.
2. By empty stack: the machine accepts if it empties the stack after processing the input.

For many textbook examples, acceptance by empty stack is particularly natural.

## Important terminology and notation

- Stack: a last-in, first-out data structure
- Push: add a symbol to the top of the stack
- Pop: remove the top symbol from the stack
- Top of stack: currently visible stack symbol
- Input head: the position in the input currently being read
- Accepting state: final state at which input is accepted
- Empty stack acceptance: accept when the stack becomes empty
- Transition: a move based on current state, input symbol, and stack top

## Why the stack matters

A DFA has no memory beyond its current state, so it cannot count arbitrarily many unmatched opening brackets or compare counts of different symbols. A PDA can push symbols onto a stack and later match them when needed.

This is the key reason PDAs are more expressive than DFAs.

## Example: language of balanced parentheses

Consider the language:

L = { ( )^n : n ≥ 0 }

This language contains strings like:

- ε
- ()
- (())
- ()()

A PDA can accept this by pushing a marker when it sees a left parenthesis and popping when it sees a right parenthesis.

If the stack is empty after processing the whole input, the number of opening and closing parentheses is balanced.


```python
# Simulating a simple PDA for balanced parentheses.
# We use the stack as a Python list.
# If the input is a balanced parentheses string, the stack will end empty.


def is_balanced_parentheses(word):
    stack = []

    for ch in word:
        if ch == '(':
            # Push an opening bracket onto the stack.
            stack.append(ch)
        elif ch == ')':
            # If there is a matching opening bracket, pop it.
            if stack and stack[-1] == '(':
                stack.pop()
            else:
                # Unmatched closing bracket: reject immediately.
                return False
        else:
            # Ignore other symbols in this simplified example.
            pass

    return len(stack) == 0


# Test examples
for w in ['', '()', '(())', '()()', '())', '((())', '(()())']:
    print(w, '->', is_balanced_parentheses(w))

print("\nThis is the basic idea behind PDA stack behavior: match opens with closes.")
```

## Example 1: simple stack behavior

For the input `(()())`, the PDA processes symbols left to right.

- read `(` → push `(`
- read `(` → push `(`
- read `)` → pop matching `(`
- read `(` → push `(`
- read `)` → pop matching `(`
- read `)` → pop matching `(`

At the end, the stack is empty, so the string is accepted.

## Example 2: rejecting an unbalanced string

For `())`, the steps are:

- `(` pushes
- `)` pops
- `)` tries to pop but the stack is empty, so reject

This shows the stack is doing real work: it keeps track of unmatched opens.

## Example 3: a more general language

A PDA can also recognize strings where the number of a's equals the number of b's, in a suitable pattern such as `a^n b^n`.

The machine pushes `A` for each `a`, then pops one `A` for each `b`. If it ends with an empty stack, the string is accepted.

This is the classic textbook example for PDA power.

## Python implementation: recognizing a^n b^n

The following function recognizes strings of the form `a^n b^n` using a stack.


```python
# PDA-like recognizer for a^n b^n
# We push 'A' for each 'a' and pop it for each 'b'.


def is_an_bn(word):
    stack = []
    seen_b = False

    for ch in word:
        if ch == 'a':
            if not seen_b:
                stack.append('A')
            else:
                # Once b's start, no more a's are allowed.
                return False
        elif ch == 'b':
            seen_b = True
            if not stack:
                return False
            stack.pop()
        else:
            return False

    return seen_b and len(stack) == 0


for w in ['aabb', 'ab', 'aaabbb', 'abb', 'aab', 'abba', '']:
    print(w, '->', is_an_bn(w))

print("\nThis demonstrates the core PDA idea: use the stack to remember unmatched a's.")
```
