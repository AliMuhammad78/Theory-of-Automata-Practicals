## Basic theory

A context-free grammar is a 4-tuple G = (V, Σ, P, S), where:

- V is the set of nonterminal symbols,
- Σ is the set of terminal symbols,
- P is the set of production rules,
- S is the start symbol.

A grammar is in Chomsky Normal Form if every production is of one of these forms:

1. A → BC, where A, B, C are nonterminal symbols
2. A → a, where a is a terminal symbol
3. S → ε, only if ε is in the language and S is the start symbol

Additionally, no other production forms are allowed.

This means that:

- there are no unit productions like A → B,
- there are no useless long productions like A → BCD,
- there are no mixed productions like A → aB or A → BCa,
- every derivation step either produces a terminal symbol or splits into two nonterminal symbols.

## Why CNF matters

CNF is useful because it standardizes grammar form. It makes proofs and algorithms cleaner, especially in parsing and formal language theory. In CNF, every derivation tree has a very regular binary structure, which is easier to reason about mathematically.

## Important terminology and notation

- CFG: context-free grammar
- Nonterminal: a variable symbol like S, A, B
- Terminal: a symbol like a, b, c from the alphabet
- Production: a rule of the form A → α
- Start symbol: the designated beginning symbol S
- Derivation: a sequence of grammar rewrites
- Nullable symbol: a symbol that can derive ε
- CNF: a restricted form of grammar used in theory and parsing

## Example grammar

Consider the grammar:

S → AS | a
A → a

This grammar generates strings of the form a^n.

However, it is not in CNF because:

- S → AS is not of the form A → BC,
- S → a is okay, but A → a is okay,
- the grammar also has a nonterminal on the left of the production and another nonterminal on the right.

To convert to CNF, we must restructure productions so that every rule uses only two nonterminals or one terminal.

## A simple CNF conversion idea

A common approach is:

1. remove ε-productions,
2. remove unit productions,
3. eliminate terminals that appear in longer productions,
4. split long productions into binary forms.

We do not always need a fully general conversion algorithm in an introductory notebook, but the important idea is that each production must be reduced to a simple form.

## Python representation of a grammar

We can represent a grammar as a dictionary where each key is a nonterminal and each value is a list of productions.


```python
# A simple context-free grammar
# This grammar generates strings of the form a^n b^n
# It is not yet in CNF.

grammar = {
    'S': ['AB', 'a'],
    'A': ['a', 'AS'],
    'B': ['b']
}

print("Grammar:")
for left, rights in grammar.items():
    print(left, '->', ' | '.join(rights))

# A production is a string that can be a terminal, a nonterminal, or a sequence.
# In CNF, we must ensure the right-hand side is either:
#   - a single terminal, OR
#   - two nonterminals
```

## Example 1: identify CNF violations

Take this grammar:

S → AB
A → a
B → bC
C → c

This grammar is not in CNF because B → bC mixes a terminal and a nonterminal.

A valid CNF version would use a new nonterminal for the terminal b, such as:

S → AB
A → a
B → X C
X → b
C → c

This is closer to CNF because the right side of each production is either a terminal or two nonterminals.

## Example 2: long productions

A production like:

A → B C D

is not allowed in CNF, because the right-hand side has length 3. We must split it into:

A → B E
E → C D

This leads to a binary derivation structure.

## Example 3: unit productions

Unit productions such as:

A → B

are not allowed in CNF unless B is a terminal, which is not the case here. Such rules must be eliminated or replaced by a more explicit derivation.

## Python implementation: checking whether a grammar is in CNF

The function below checks the basic CNF condition for a grammar stored as a dictionary.


```python
# Check whether a grammar is in basic Chomsky Normal Form.
# This version checks a simplified condition:
# - each rule is either terminal or two nonterminals,
# - no epsilon rule is included unless it is the start rule.

def is_cnf(grammar):
    nonterminals = set(grammar.keys())

    for left, productions in grammar.items():
        for prod in productions:
            if prod == '':
                if left != 'S':
                    return False
                continue

            if len(prod) == 1 and prod.islower():
                continue

            if len(prod) == 2 and prod[0].isupper() and prod[1].isupper():
                continue

            return False

    return True


cnf_grammar = {
    'S': ['AB', 'a'],
    'A': ['a'],
    'B': ['b']
}

not_cnf_grammar = {
    'S': ['AB', 'a'],
    'A': ['a', 'AS'],
    'B': ['b']
}

print("cnf_grammar is CNF:", is_cnf(cnf_grammar))
print("not_cnf_grammar is CNF:", is_cnf(not_cnf_grammar))

# This function is deliberately simple and educational.
# It demonstrates the exact structural rule that defines CNF.
```

## Final summary

Chomsky Normal Form is a restricted form of context-free grammar in which every production is either a binary nonterminal rule or a single terminal rule. The main idea is to make grammatical structure uniform and easier to analyze mathematically. In practice, converting a grammar to CNF involves removing epsilon rules, removing unit productions, replacing terminals in longer productions, and splitting long rules into binary ones.

This gives us a regular, structured grammar that is easier to reason about in formal language theory.
